```python
from pathlib import Path

test_images_dir = Path("C:/Users/SANMAYA/Downloads/archive2_unzipped/css-data/test/images")

files = sorted([p.name for p in test_images_dir.glob("*") if p.suffix.lower() in [".jpg",".jpeg",".png"]])
print("Found", len(files), "test images. First 10:")
for f in files[:10]:
    print(" ", f)

# choose the first image automatically
SAMPLE_IMG = files[0]
print("\nUsing SAMPLE_IMG:", SAMPLE_IMG)
```

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# Load the best trained model
BEST_MODEL = "runs/detect/construction_ppe_model/weights/best.pt"
model = YOLO(BEST_MODEL)

# Use the sample image you selected earlier
img_path = test_images_dir / SAMPLE_IMG
print("Testing image:", img_path)

# Run inference
results = model.predict(
    source=str(img_path),
    conf=0.35,   # confidence threshold
    iou=0.45,    # IoU threshold
    imgsz=640,
)

# Annotated image output
annotated_img = results[0].plot()

# Convert BGR → RGB for displaying
annotated_img = cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB)

# Show image
plt.figure(figsize=(10, 10))
plt.imshow(annotated_img)
plt.axis("off")
plt.show()
```
